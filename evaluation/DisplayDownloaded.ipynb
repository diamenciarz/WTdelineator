# Display downloaded / processed coverage

How many records from each catalogue are already in `data/evaluation/processed/`.

STAFF III also shows how many compressed WFDB files are still sitting in `data/staff_III` (the shrinking cache).

```python
from pathlib import Path
import sys

REPO = Path.cwd() if (Path.cwd() / "evaluation" / "common.py").exists() else Path.cwd().parent
sys.path.insert(0, str(REPO / "evaluation"))

import pandas as pd
from IPython.display import display

import common as C

inv = C.dataset_inventory()
inv_show = inv.copy()
inv_show["percent"] = (inv_show["fraction"] * 100).round(2)
inv_show["processed_MB"] = (inv_show["processed_bytes"] / 1e6).round(1)
inv_show["projected_processed_size"] = (inv_show["projected_processed_bytes"] / 1e6).round(1)
display(inv_show[["title", "processed", "catalogue", "percent", "cache_wfdb", "processed_MB", "projected_processed_size"]])
_ = C.plot_download_fractions(inv)
```

```python
staff_cache = C.list_cache_record_ids("staff_iii")
print(f"STAFF III WFDB still in data/staff_III: {len(staff_cache)}")
print(f"LUDB WFDB leftover: {len(C.list_cache_record_ids('ludb'))}")
print(f"QTDB WFDB leftover: {len(C.list_cache_record_ids('qtdb'))}")
print(f"PTB-XL WFDB leftover: {len(C.list_cache_record_ids('ptb_xl'))}")
```
