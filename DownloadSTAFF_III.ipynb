```python
"""
Resumable STAFF III Dataset Downloader
Downloads to: data/staff_III/
Handles interruptions gracefully and continues from where it left off.
"""

import os
import wfdb
from tqdm import tqdm
import time
```

# CONFIG

```python
DATASET_NAME = "staffiii"
BASE_DIR = "data/staff_III"
os.makedirs(BASE_DIR, exist_ok=True)

# Optional: limit to specific records (None = all)
# Example: RECORD_LIST = ["001a", "002b", ...]
RECORD_LIST = None
```

```python

def get_existing_records(download_dir: str) -> set:
    """Return set of already downloaded record names (based on .hea files)"""
    existing = set()
    for file in os.listdir(download_dir+"/data"):
        if file.endswith('.hea'):
            record_name = file[:-4]  # remove .hea
            existing.add("data/"+record_name)
    return existing


def download_staff_iii():
    print(f"Starting/resuming download of STAFF III dataset to: {BASE_DIR}\n")
    
    # Get list of all records
    print("Fetching list of records...")
    all_records = wfdb.get_record_list(DATASET_NAME)
    print(f"Total records found: {len(all_records)}")
    
    # Filter already downloaded records
    existing = get_existing_records(BASE_DIR)
    print(f"Already downloaded: {len(existing)} records")
    
    to_download = [r for r in all_records if r not in existing]
    
    if not to_download:
        print("✅ All records are already downloaded!")
        return
    
    print(f"Records remaining to download: {len(to_download)}\n")
    
    # Download one by one with progress
    for i, record_name in enumerate(tqdm(to_download, desc="Downloading records")):
        try:
            wfdb.dl_files(
                db=DATASET_NAME,
                dl_dir=BASE_DIR,
                files=[f"{record_name}.dat", f"{record_name}.hea"]  # main files
            )
            
            # Optional: also download annotation files if they exist
            wfdb.dl_files(db=DATASET_NAME, dl_dir=BASE_DIR, files=[f"{record_name}.atr"])
            
        except Exception as e:
            continue
        
    print("\n🎉 Download process completed!")
    final_count = len(get_existing_records(BASE_DIR))
    print(f"Total records now available: {final_count}/{len(all_records)}")
```
